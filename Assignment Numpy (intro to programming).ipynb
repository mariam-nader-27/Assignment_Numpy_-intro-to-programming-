```python
arr = np.random.randint(1, 100, size=(3, 3))
max1=arr.max()
min1=arr.min()
formula = (arr - min1) / (max1 - min1)
print(formula)
```

Question 2:

```python
arr = np.random.randint(1, 100, size=10)
x = float(input("enter a number: "))
y=np.abs(arr - x)
closest_value = arr[y.argmin()]
print("array:\n",arr)
print("you enter:",x)
print("the closest value for your number is:\n",closest_value)
```

```python
arr = np.random.randint(1, 100, size=(6, 4))
row_sums = arr.sum(axis=1)
even_rows = np.where(row_sums % 2 == 0)
print("array:\n",arr)
print("the sum of rows:\n",row_sums)
print("the index of even rows:\n",even_rows)
```
